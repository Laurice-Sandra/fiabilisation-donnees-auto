```python
import os, sys

os.environ["JAVA_HOME"] = os.path.join(sys.prefix, "Library")
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
os.environ["SPARK_LOCAL_HOSTNAME"] = "localhost"   # remplace LAURY_MUSE par un nom valide

from pyspark.sql import SparkSession
spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("verification")
    .config("spark.driver.host", "127.0.0.1")         # adresse utilisée par Spark
    .config("spark.driver.bindAddress", "127.0.0.1")  # adresse sur laquelle Spark écoute
    .getOrCreate()
)
print(spark.version)
spark.range(5).show()
```
